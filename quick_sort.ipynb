```python
def quick_sort(arr):
  # base case
  if len(arr) <= 1:
    return arr
  # recursive part
  # pick pivot
  pivot = med_of_three(arr)
  # split into two lists
  less = [x for x in arr if x < pivot]
  equal = [x for x in arr if x == pivot]
  greater = [x for x in arr if x > pivot]
  #sort smaller lists and put them all back together
  return quick_sort(less) + equal + quick_sort(greater)

def med_of_three(arr):
  first = arr[0]
  last = arr[-1]
  middle = arr[len(arr) // 2]
  if first > middle:
    first, middle = middle, first
  if middle > last:
    middle, last = last, middle
  if first > middle:
    first, middle = middle, first
  return middle


```

```python
quick_sort([9,8,77,6,4,4])
```
