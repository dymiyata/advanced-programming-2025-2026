# Dictionaries

Recall that Hash Tables are implemented in Python as Dictionaries.

You should think of Hash tables/dictionaries as arrays where you don't use an *index* to access a value, rather you use a *key* to access a value. Keys are typically strings, although you can use any immutable type (strings, integers, floats, booleans, etc...).  

We create a dictionary using `{}` and listing the key-value pairs within these braces.

````python
people = {"Chenhao" : 5, "Andy" : 19, "Aneesh" : 3, "Evan" : 7, "Varun" : 23}
````

````python
people["Varun"]
````

````python
dict2 = {1:1, 2:2, 3:5, 10:20}
dict2[3]
````

We access values in a dictionary with the syntax `dict[key]`.  Notice this is like accessing a value in an array, but instead of passing an index, you pass a key.

We can also use this syntax to create a new entry or update an existing entry in the dictionary.

````python
people["Dane"] = 57
````

````python
people["Dane"]
````

What happens if we try to access a value with key that's not already in the dictionary?

````python
people["Albert"]
````

Thus, if we're unsure if a key is in the dictionary, we need to check if it is using `if key in dict:`.  

Notice for an array, this check would be $O(n)$ time complexity, but for a dictionary/hash table this is just $O(1)$.

````python
if "Albert" in people:
  people["Albert"]
````

There is also a helpful function that combines checking if the key is in the dictionary and accessing the value for that key.  

To avoid errors when you are unsure if a key is already in your dictionary, then you should use `dict.get(key, default value)` to access the value.  
- This function will return the value for the given key if it exists.
- If the key doesn't exist it will return the `default_value` if it's specified or `None` if no default value is specified.
This is especially useful if you're using the values in a dictionary to count something. It's quite common to have lines like:
```
dict[key] = dict.get(key, 0) += 1
```
Can you understand what that line of code is doing?

````python
people.get("Dalston", 24)
print(people)
````

````python
people["Dalston"] = 24
print(people)
````

To remove a key-value pair from a dictionary, there are a few options and they're basically the same as for removing items from an array..
- `del dict[key]`
- `dict.pop(key)`

````python
del people["Dane"]
print(people)
````

````python
people.pop("Andy")
````

````python
print(people)
````

In Python you can iterate through a list using

`for item in list:`

Similarly you can iterate through the keys in a dictionary using

`for key in dictionary:`

````python
for person in people:
  print(person, people[person])
````

A few other useful methods:
- `dict.keys()`
- `dict.values()`
- `dict.items()`

````python
people_list = list(people.keys())
````

````python
people.items()
````

### Example Problem

Write a function that takes a string as input and returns a dictionary mapping each character to how many times it appears in the string. For example, if you input the string `"hello"` the function should output the dictionary
```
{"h": 1, "e": 1, "l": 2, "o": 1}
```

````python
def count_chars(s):
  my_dict = {}
  for char in s:
    if char != " ": # ignore spaces
    # if char in my_dict:
    #   my_dict[char] += 1
    # else:
    #   my_dict[char] = 1
      my_dict[char] = my_dict.get(char, 0) + 1
  return my_dict
````

````python
count_chars("hello world!")
````

## Sets

Sets are another slightly different object in Python that is also implemented as a hash table.  You should think of a set as a dictionary with keys, but no values.  

Mathematically, a set is simply a collection of objects.  Two sets are equal if every object in one is also in the other.  Thus, in a set:
- There isn't any inherent order
- Duplicating an object doesn't change the set (i.e. sets don't have duplicates).

Notice with a hash table, we can't duplicate keys (because the hash function is a *function* afterall so the same key will always give the same output). We can use hash tables to check if a key exists in $O(1)$ time.

Thus hash tables are the perfect way to implement a set. Sets are very useful when you frequently need to check if an element is present in a collection but you don't care about the order.

We can define a set using `{}` just like a dictionary.  However, instead of listing key-value pairs, we just list the keys (which are the items of the set) with no values.

````python
my_set = {1, 2, 3}
print(my_set)
````

We can also turn a list into a set using `set(list)`.  Notice this will "get rid of" the duplicate elements of the list as well as the inherent order of the list.

````python
set2 = set([2,3,4,4])
print(set2)
````

Importantly, when you define an empty dictionary we can do `dict = {}`.  However, even though we use curly braces to define sets, if we don't list any elements in the braces, then Python will assume that it's a dictionary insted.  Therefore, to define an empty set, we use `my_set = set()`.

````python
empty_set = set()
empty_dict = {}

print(type(empty_set))
print(type(empty_dict))
````

We can add or remove elements using `my_set.add()` and `my_set.discard()`.

````python
set2.add(13)
print(set2)
````

````python
set2.discard(13)
print(set2)
````

We can check if an element is in a set just like we do with lists or dictionaries.

````python
15 in set2
````

We can do the usual set operations (intersection, union, set difference) as follows

````python
print(my_set & set2) # intersection (and)
print(my_set | set2) # union (or)
````

You can get the size of a set using `len()` just like with lists.

````python
len(set2)
````

### Example Problem

Write a function that takes a list as input and returns the number of distinct elements in it.  For example if you input

`[1, 2, 3, 1, 4, 2]`

the function should return `4`.

````python
def num_distinct(arr):
  my_set = set(arr)
  return len(my_set)

num_distinct([1, 2,3,1,4,2])
````
