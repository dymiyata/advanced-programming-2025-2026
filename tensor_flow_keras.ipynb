```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from keras import layers

tf.random.set_seed(2026)
np.random.seed(2026)
keras.utils.set_random_seed(2026)
```

```python
### Creating Artificial Data

m = 2000

x1 = np.random.randint(-5, 5, m)
x2 = np.random.randint(-5, 5, m)
x3 = np.random.randint(-5, 5, m)

X = np.column_stack([x1, x2, x3])

score = x1**2 + x2 - 2*x3
y = (score > 8).astype(np.float32)
```

```python
print(X[0:10,:])
print(y[0:10])
```

```python
X.shape
```

```python
y.shape
```

```python
model = keras.Sequential([
    layers.Input(shape=(3,)),
    layers.Dense(4, activation="relu"),
    layers.Dense(4, activation="relu"),
    layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(
    optimizer="adam",
    loss="crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(X, y, epochs=100, verbose=False)
```

```python
model.evaluate(X, y)
```
