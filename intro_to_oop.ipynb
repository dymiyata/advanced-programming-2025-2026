# Object Oriented Programming


Example: Students class
- attributes: name and age

```python
class Student:
  ## __init__ function
  # this runs whenever you create an object of this class
  # this is the function that stores the attributes of the class
  def __init__(self, name, age, grade):
    # Attributes are name, age, and grade
    self.name = name
    self.age = age
    self.grade = grade

  # Now we can define methods
  def introduce(self):
    print(f"Hi my name is  {self.name}, I'm {self.age} years old and I'm in the {self.grade} grade!")
```

Let's make a Student object:

```python
a = Student("Aneesh", 12, 7)
```

Access attributes of `a`

```python
a.introduce()
```

```python
b = Student("Andy", 13, 8)
```

```python
b.introduce()
```

```python
introduce()
```

Let's define a `Counter` class.
- want an attribute to keep track of the current value
- want a method to increment the counter
- want a method to reset the counter

```python
class Counter:
  def __init__(self, starting_value = 0):
    self.value = starting_value

  def increment(self):
    self.value += 1

  def reset(self):
    self.value = 0
```

```python
counter = Counter()
```

```python
counter.value
```

```python
counter.increment()
```

```python
counter.value
```

```python
for i in range(5):
  counter.increment()
counter.value
```

```python
counter.reset()
counter.value
```
