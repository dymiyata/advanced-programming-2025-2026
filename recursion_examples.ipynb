# Recursion Examples

### Example 1

Write a function `triang(n)` which takes a positive integer `n` as input, and returns the sum of the integers from `1` to `n`.

````python
### Non recursive version
def triang(n):
  sum = 0
  for x in range(n):
    sum += x + 1
  return sum


### Recursive version
def recursive_triang(n):
  if n == 1:
    return 1
  return recursive_triang(n-1) + n



````

````python
recursive_triang(5)
````

### Example 2

Write a function `count_down(n)` which takes a positive integer `n` as input, and prints the integers from `1` to `n` but counting down.  For example `count_down(4)` should print:

```
4
3
2
1
```
The function doesn't need to return anything.

````python
def count_down(n):
  if n == 1:
    print(1)
  else:
    print(n)
    count_down(n-1)
````

````python
count_down(4)
````

### Example 3

Write a function `count_paths(m, n)` which takes two positive integers `m` and `n` as input.  

It should return the number of paths to go from the bottom left to the top right of an $m \times n$ grid where you can only move 1 unit right or 1 unit up at each step.



````python
from functools import lru_cache # memoization

@lru_cache
def count_paths(m, n):
  # base case
  if m == 1:
    return n + 1
  if n == 1:
    return m + 1
  return count_paths(m, n - 1) + count_paths(m - 1, n)
````

````python
count_paths(100, 100)
````
