We want to get the computer to do propositional logic

```python
class Atomic:
  def __init__(self, name):
    self.name = name

  def evaluate(self, model): # evaluate truth value of this proposition
    # model is a row of the truth table
    # model will be a dict: {"P": True, "Q": False, ...}
    return model[self.name]

  def symbols(self): # gets all atomic propositions this one is built out of
    return {self.name}

  def formula(self): # return a string to print the proposition "nicely"
    return self.name
```

```python
class Or:
  def __init__(self, *props):
    # left and right are propositions
    self.props = list(props)

  def evaluate(self, model):
    for prop in self.props:
      if prop.evaluate(model):
        return True
    return False

  def symbols(self):
    s = set()
    for prop in self.props:
      s = s | prop.symbols()
    return s
    # return self.left.symbols().union(self.right.symbols())

  def formula(self):
    text = " \u2228 ".join([prop.formula() for prop in self.props])
    return f"({text})"
```

```python
class Not:
  def __init__(self, prop):
    self.prop = prop

  def evaluate(self, model):
    return not self.prop.evaluate(model)

  def formula(self):
    return f"\u00AC{self.prop.formula()}"

  def symbols(self):
    return self.prop.symbols()
```

```python
class Conditional:
  def __init__(self, left, right):
    self.left = left
    self.right = right

  def evaluate(self, model):
    if not self.left.evaluate(model): # if the "if" part is false, return True
      return True
    elif self.right.evaluate(model): # else if the "then part is true, return True"
      return True
    else:
      return False # in all other cases, return False
    # one line way to do it remembering that "if P, then Q" is the same as "(not P) or Q"
    return (not self.left.evaluate(model)) or self.right.evaluate(model)

  def formula(self):
    return f"({self.left.formula()} \u2192 {self.right.formula()})"

  def symbols(self):
    return self.left.symbols() | self.right.symbols()
```

```python
from itertools import product

def all_models(symbols): # Generates all possible models ("rows of truth table")
  symbols = list(symbols)

  for values in product([True, False], repeat=len(symbols)):
    model = {}
    for i in range(len(symbols)):
      model[symbols[i]] = values[i]
    yield model


def entails(kb, query):
  all_symbols = kb.symbols() | query.symbols()
  for model in all_models(all_symbols):
    if kb.evaluate(model) and (not query.evaluate(model)): # check if kb is true but query is false
      return False
  return True


```
