```python
# List (array) implementation
class Queue:
  def __init__(self):
    self.data = []

  def enqueue(self, item): # O(n) in the worse case
    self.data.append(item)

  def dequeue(self): # Always O(n)
    return self.data.pop(0)
```

```python
q = Queue()
```

```python
q.enqueue(3)
```

```python
for i in range(10):
  q.enqueue(i)
```

```python
q.dequeue()
```

```python
q.dequeue()
```

```python
q.dequeue()
```

```python
# List (array) implementation
class Stack:
  def __init__(self):
    self.data = []

  def push(self, item): # O(n) in the worst case
    self.data.append(item)

  def pop(self): # O(1)
    return self.data.pop()
```

```python
s = Stack()
for i in range(10):
  s.push(i)
```

```python
s.pop()
```

```python
s.pop()
```

```python
# Linked list implementations
# deque means "double ended queue" (doubly linked list)
# A doubly linked list is a list with pointers in both directions
from collections import deque

class Queue:
  def __init__(self):
    self.data = deque() # Empty deque

  def enqueue(self, item): # O(1)
    self.data.append(item)

  def dequeue(self): # O(1)
    return self.data.popleft() # remove from start of linked list


class Stack:
  def __init__(self):
    self.data = deque()

  def push(self, item): # O(1)
    self.data.append(item)

  def pop(self): # O(1)
    return self.data.popright()
```

```python
q2 = Queue()
```

```python
q2.enqueue(3)
```

```python
q2.dequeue()
```
